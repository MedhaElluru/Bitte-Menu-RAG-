```python
#Programmatic way to upload files and store them in a vector store
#files --> embedding --> vector store

files_path = ['Bitte - Apresentação Corporativa.pdf',
             'Termos e Condições _ Politica de Privacidade da Bitte (PT_EN).docx' ]
files_id = []

#upload files to OpenAI API

for file_path in files_path:
  with open(file_path, 'rb') as file:
    response = client.files.create(file = file, purpose = 'assistants')
    files_id.append(response.id)

files_id
```

```python
#creating vector store and add the uploaded files
vector_store = client.vector_stores.create(
    name = 'Bitte-assistant'
)
vector_store_id = vector_store.id
vector_store_id
```

```python
for file_id in files_id:
  client.vector_stores.files.create(
      vector_store_id = vector_store_id,
      file_id = file_id
  )

  #embeddings happen here automatically
```
